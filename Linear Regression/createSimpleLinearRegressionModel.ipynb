```python
class SelfLinearRegression:

    def __init__(self):
        self.m = 0
        self.b = 0

    def fit(self, x_train, y_train):

        print(x_train)

        numerator = 0
        denominator = 0

        print(x_train)

        for i in range(len(x_train)):
            
            print("Numerator =", numerator,
                  "And Denominator =", denominator)

            numerator = numerator + (
                (x_train.iloc[i] - x_train.mean()) *
                (y_train.iloc[i] - y_train.mean())
            )

            denominator = denominator + (
                (x_train.iloc[i] - x_train.mean()) *
                (x_train.iloc[i] - x_train.mean())
            )

        self.m = numerator / denominator

        self.b = (
            y_train.mean() -
            (self.m * x_train.mean())
        )

    def predict(self, x_test):

        print(
            "The value of self.b:", self.b,
            "And The value of self.m:", self.m
        )

        return self.b + self.m * x_test
```
