```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/kaggle/input/datasets/syedanwarafridi/vehicle-sales-data/car_prices.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(16, 5))

plt.subplot(1, 2, 1)
sns.histplot(df["odometer"], kde=True)

plt.title("Distribution of Odometer")
plt.xlabel("Odometer")

plt.show()
```

```python
df.describe()
```

```python
sns.boxplot(df['odometer'])
```

```python
percentile25=df['odometer'].quantile(0.25)
percentile75=df['odometer'].quantile(0.75)
```

```python
percentile25
```

```python
percentile75
```

```python
iqr=percentile75-percentile25
```

```python
iqr
```

```python
upper_limit=percentile75+1.5*iqr
lower_limit=percentile25-1.5*iqr
```

```python
print("Upper limit:",upper_limit)
print("Lower limit:",lower_limit)
```

```python
df[df['odometer']<upper_limit]
```

# **Finding Outliers**

```python
df[df['odometer']>upper_limit]
```

```python
df[df['odometer']<lower_limit]
```

```python
new_df=df[df['odometer']<upper_limit]
```

```python
new_df.shape
```

```python
plt.figure(figsize=(16,8))
plt.subplot(2,2,1)
sns.histplot(df['odometer'])

plt.subplot(2,2,2)
sns.boxplot(df['odometer'])

plt.subplot(2,2,3)
sns.histplot(new_df['odometer'])

plt.subplot(2,2,4)
sns.boxplot(new_df['odometer'])

plt.show()
```

```python
# from sklearn.preprocessing import StandarScaler

# scale=StandarScaler()
# scale.fit_transform(df[['odometer

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df['odometer'] = scaler.fit_transform(df[['odometer']])

newScaler=StandardScaler()
new_df['odometer']=newScaler.fit_transform(new_df[['odometer']])
```

```python
plt.figure(figsize=(16,8))
plt.subplot(2,2,1)
sns.histplot(df['odometer'])

plt.subplot(2,2,2)
sns.boxplot(df['odometer'])

plt.subplot(2,2,3)
sns.histplot(new_df['odometer'])

plt.subplot(2,2,4)
sns.boxplot(new_df['odometer'])

plt.show()
```

# *Capping*****

```python
new_df_cap = df.copy()

new_df_cap['odometer'] = np.where(
    new_df_cap['odometer'] > upper_limit,
    upper_limit,
    np.where(
        new_df_cap['odometer'] < lower_limit,
        lower_limit,
        new_df_cap['odometer']
    )
)
```
